```python
import numpy as np
import matplotlib.pyplot as plt
import tensorflow as tf
from tensorflow import keras
from tensorflow.keras import layers
```

```python
(x_train, y_train), (x_test, y_test) = keras.datasets.mnist.load_data()

print("Training data:", x_train.shape)
print("Testing data:", x_test.shape)
```

```python
#normalize dataset
x_train = x_train.astype('float32') / 255.0
x_test = x_test.astype('float32') / 255.0
```

```python
#Select normal and anomaly data Selects only digit 0 images for training Selects digits 1–9 as possible anomalies.
x_train_normal = x_train[y_train == 0]

x_test_normal = x_test[y_test == 0]
x_test_anomaly = x_test[y_test != 0]

print("Normal training images:", x_train_normal.shape)
print("Normal testing images:", x_test_normal.shape)
print("Anomaly testing images:", x_test_anomaly.shape)
```

```python
plt.figure(figsize=(10, 3))

for i in range(5):
    plt.subplot(1, 5, i + 1)
    plt.imshow(x_train_normal[i], cmap='gray')
    plt.axis('off')

plt.show()
```

```python
#Encoder
input_img = keras.Input(shape=(28, 28))

x = layers.Flatten()(input_img)
x = layers.Dense(64, activation='relu')(x)
x = layers.Dense(32, activation='relu')(x)

encoded = layers.Dense(16, activation='relu')(x)

encoder = keras.Model(input_img, encoded)
```

```python
#decoder
latent_input = keras.Input(shape=(16,))

x = layers.Dense(32, activation='relu')(latent_input)
x = layers.Dense(64, activation='relu')(x)
x = layers.Dense(784, activation='sigmoid')(x)

decoded = layers.Reshape((28, 28))(x)

decoder = keras.Model(latent_input, decoded)
```

```python
#combine
autoencoder_input = keras.Input(shape=(28, 28))

encoded_image = encoder(autoencoder_input)
decoded_image = decoder(encoded_image)

autoencoder = keras.Model(autoencoder_input, decoded_image)
```

```python
#compile model
autoencoder.compile(
    optimizer='adam',
    loss='mse',
    metrics=['mae']
)
```

```python
#train auto
history = autoencoder.fit(
    x_train_normal,
    x_train_normal,
    epochs=10,
    batch_size=128,
    validation_split=0.1
)
```

```python
plt.plot(history.history['loss'], label='Training Loss')
plt.plot(history.history['val_loss'], label='Validation Loss')

plt.title('Autoencoder Training Loss')
plt.xlabel('Epochs')
plt.ylabel('Loss')
plt.legend()
plt.show()
```

```python
train_predictions = autoencoder.predict(x_train_normal)

train_error = np.mean(
    np.square(x_train_normal - train_predictions),
    axis=(1, 2)
)

threshold = np.percentile(train_error, 95)

print("Anomaly Threshold:", threshold)
```

```python
test_images = np.concatenate([
    x_test_normal[:100],
    x_test_anomaly[:100]
])

test_predictions = autoencoder.predict(test_images)

test_error = np.mean(
    np.square(test_images - test_predictions),
    axis=(1, 2)
)

results = test_error > threshold

print("Total Images:", len(test_images))
print("Detected Anomalies:", np.sum(results))
print("Detected Normal Images:", np.sum(~results))
```

```python
plt.figure(figsize=(10, 4))

for i in range(5):
    plt.subplot(2, 5, i + 1)
    plt.imshow(test_images[i], cmap='gray')
    plt.title("Original")
    plt.axis('off')

    plt.subplot(2, 5, i + 6)
    plt.imshow(test_predictions[i], cmap='gray')
    plt.title("Reconstructed")
    plt.axis('off')

plt.show()
```
